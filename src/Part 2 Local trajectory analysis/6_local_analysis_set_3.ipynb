```python
def min_max_normalization(value):
    """归一化，也称离差标准化
    公式：(原始值-最小值)/(最大值-最小值)
    :return 值域[0,1]
    """
    new_value = (value - value.min()) / (value.max() - value.min())
    return new_value
```
